## mem-readwrite (RAM read&write)

```python
import pandas as pd
import numpy as np
from pathlib import Path

ransomware = ["Conti", "Darkside", "LockBit", "REvil", "Ryuk", "WannaCry"]
benign = ["AESCrypt", "Firefox", "Idle", "Office", "SDelete", "Zip"]

data_root = Path("/kaggle/input/datasets/hiranomanabu/ransmap-2024-ransomware-behavioral-features/RanSMAP/dataset")
```

```python
from validation_ruless import check_files, check_type, check_size,check_entropy,check_gpa
```

```python

benign_mem_readwrite_paths = []
ransomware_mem_readwrite_paths = []

for folder in data_root.iterdir():
    if not folder.is_dir():
        continue

    for csv_path in folder.rglob("*.csv"):
        if "mem" in csv_path.name.lower() and "readwrite" in csv_path.name.lower():
            for part in csv_path.parts:
                if part in ransomware:
                    ransomware_mem_readwrite_paths.append(csv_path)
                    break
                elif part in benign:
                    benign_mem_readwrite_paths.append(csv_path)
                    break

print("Ransomware mem-readwrite files:", len(ransomware_mem_readwrite_paths))
print("Benign mem-readwrite files:", len(benign_mem_readwrite_paths))
```

```python
check_files(ransomware_mem_readwrite_paths,0,1)
print("Ransomware_mem_readwrite checking done")
check_files(benign_mem_readwrite_paths,0,1)
print("Benign_mem_readwrite checking done")


print("Checking unix timestamp in sec and nsec is done successfully ! ")
```

#### Checking type : 

```python
check_type(ransomware_mem_readwrite_paths,5)
print("Ransomware_mem_readwrite checking done")
check_type(benign_mem_readwrite_paths,5)
print("Benign_mem_readwrite checking done")


print("Checking types is done successfully ! ")
```

#### Checking size : 

```python
check_size(ransomware_mem_readwrite_paths,3)
print("Ransomware_mem_readwrite checking done")
check_size(benign_mem_readwrite_paths,3)
print("Benign_mem_readwrite checking done")

print("Checking size is done successfully!")
```

```python
check_entropy(ransomware_mem_readwrite_paths,4)
print("Ransomware_mem_readwrite checking done")
check_entropy(benign_mem_readwrite_paths,4)
print("Benign_mem_readwrite checking done")
print("Checking entropy is done successfully !")
```
