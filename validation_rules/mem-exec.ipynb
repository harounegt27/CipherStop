## mem-exce (RAM excecution)

```python
import pandas as pd
import numpy as np
from pathlib import Path

ransomware = ["Conti", "Darkside", "LockBit", "REvil", "Ryuk", "WannaCry"]
benign = ["AESCrypt", "Firefox", "Idle", "Office", "SDelete", "Zip"]

data_root = Path("/kaggle/input/datasets/hiranomanabu/ransmap-2024-ransomware-behavioral-features/RanSMAP/dataset")
```

```python
from validation_ruless import check_files, check_type,check_gpa
```

```python

benign_mem_exec_paths = []
ransomware_mem_exec_paths = []

for folder in data_root.iterdir():
    if not folder.is_dir():
        continue

    for csv_path in folder.rglob("*.csv"):
        if "mem" in csv_path.name.lower() and "exec" in csv_path.name.lower():
            for part in csv_path.parts:
                if part in ransomware:
                    ransomware_mem_exec_paths.append(csv_path)
                    break
                elif part in benign:
                    benign_mem_exec_paths.append(csv_path)
                    break

print("Ransomware mem-exec files:", len(ransomware_mem_exec_paths))
print("Benign mem-exec files:", len(benign_mem_exec_paths))
```

```python
check_files(ransomware_mem_exec_paths,0,1)
print("Ransomware_mem_rexec checking done")
check_files(benign_mem_exec_paths,0,1)
print("Benign_mem_exec checking done")


print("Checking unix timestamp in sec and nsec is done successfully ! ")
```

#### Checking type : 

```python
check_type(ransomware_mem_exec_paths,5)
print("Ransomware_mem_exec checking done")
check_type(benign_mem_exec_paths,5)
print("Benign_mem_exec checking done")


print("Checking types is done successfully ! ")
```
