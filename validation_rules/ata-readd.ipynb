## ata-read (storage read)

```python
import pandas as pd
import numpy as np
from pathlib import Path

ransomware = ["Conti", "Darkside", "LockBit", "REvil", "Ryuk", "WannaCry"]
benign = ["AESCrypt", "Firefox", "Idle", "Office", "SDelete", "Zip"]

data_root = Path("/kaggle/input/datasets/hiranomanabu/ransmap-2024-ransomware-behavioral-features/RanSMAP/dataset")
```

```python
from validation_ruless import check_files, check_lba, check_size
```

```python

benign_ata_read_paths = []
ransomware_ata_read_paths = []

for folder in data_root.iterdir():
    if not folder.is_dir():
        continue

    for csv_path in folder.rglob("*.csv"):
        if "ata" in csv_path.name.lower() and "read" in csv_path.name.lower():
            for part in csv_path.parts:
                if part in ransomware:
                    ransomware_ata_read_paths.append(csv_path)
                    break
                elif part in benign:
                    benign_ata_read_paths.append(csv_path)
                    break

print("Ransomware ATA-read files:", len(ransomware_ata_read_paths))
print("Benign ATA-read files:", len(benign_ata_read_paths))
```

```python
print(ransomware_ata_read_paths[10])
```

#### Checking timestamps : 

```python
check_files(ransomware_ata_read_paths,0,1)
print("Ransomware_ata_read checking done")
check_files(benign_ata_read_paths,0,1)
print("Benign_ata_read checking done")


print("Checking unix timestamp in sec and nsec is done successfully ! ")
```

#### Checking lba : 

```python
check_lba(ransomware_ata_read_paths,2)
print("Ransomware_ata_read checking done")
check_lba(benign_ata_read_paths,2)
print("Benign_ata_read checking done")

print("Checking LBA is done successfully ! ")
        
```

#### Checking size : 

```python
check_size(ransomware_ata_read_paths,3)
print("Ransomware_ata_read checking done")
check_size(benign_ata_read_paths,3)
print("Benign_ata_read checking done")

print("Checking size is done successfully!")
```
